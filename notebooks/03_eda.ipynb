```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
```

```python
df = pd.read_csv(
    "../data/processed/hourly_cleaned.csv"
)
```

```python
machine_production = (
    df.groupby("equipment_ID")[
        "%production"
    ]
    .mean()
    .reset_index()
)

machine_production
```

```python
fig = px.bar(
    machine_production,
    x="equipment_ID",
    y="%production",
    title="Average Production by Machine"
)

fig.show()
```

```python
machine_downtime = (
    df.groupby("equipment_ID")[
        "%downtime"
    ]
    .mean()
    .reset_index()
)

fig = px.bar(
    machine_downtime,
    x="equipment_ID",
    y="%downtime",
    title="Average Downtime by Machine"
)

fig.show()
```
