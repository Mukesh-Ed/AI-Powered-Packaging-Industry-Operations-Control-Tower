```python
import pandas as pd
import numpy as np
```

```python
raw_data = pd.read_csv(
    "../data/raw/raw_data.csv"
)

hourly_data = pd.read_csv(
    "../data/raw/sequences_1h_data.csv"
)
```

```python
raw_data = raw_data.drop_duplicates()

hourly_data = hourly_data.drop_duplicates()

print(raw_data.shape)
print(hourly_data.shape)
```

```python
raw_data["interval_start"] = pd.to_datetime(
    raw_data["interval_start"],
    errors="coerce"
)

hourly_data["interval_start"] = pd.to_datetime(
    hourly_data["interval_start"],
    errors="coerce"
)
```

```python
numeric_columns = [
    "elapsed",
    "pi",
    "po",
    "speed"
]

for column in numeric_columns:

    if column in raw_data.columns:

        raw_data[column] = pd.to_numeric(
            raw_data[column],
            errors="coerce"
        )
```

```python
for column in hourly_data.columns:

    if column not in [
        "equipment_ID",
        "interval_start"
    ]:

        hourly_data[column] = pd.to_numeric(
            hourly_data[column],
            errors="coerce"
        )
```

```python
print(
    raw_data.isnull().sum()
    .sort_values(ascending=False)
)
```

```python
numeric_cols = raw_data.select_dtypes(
    include=np.number
).columns

raw_data[numeric_cols] = (
    raw_data[numeric_cols]
    .fillna(0)
)
```

```python
raw_data = raw_data.dropna(
    subset=["equipment_ID"]
)

hourly_data = hourly_data.dropna(
    subset=["equipment_ID"]
)
```

```python
raw_data.to_csv(
    "../data/processed/raw_cleaned.csv",
    index=False
)

hourly_data.to_csv(
    "../data/processed/hourly_cleaned.csv",
    index=False
)
```
